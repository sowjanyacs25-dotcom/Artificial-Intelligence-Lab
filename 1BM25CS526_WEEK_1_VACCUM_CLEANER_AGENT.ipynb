```python
def vacuum_agent(location, room_status):
    if room_status[location] == "Dirty":
        return "Suck"
    elif location == "A":
        return "Move Right"
    elif location == "B":
        return "Move Left"

room_status = {
    "A": "Dirty",
    "B": "Dirty"
}

location = "A"

for step in range(6):
    print(f"\nStep {step + 1}")
    print("Location:", location)
    print("Room status:", room_status)

    action = vacuum_agent(location, room_status)
    print("Action:", action)

    if action == "Suck":
        room_status[location] = "Clean"
    elif action == "Move Right":
        location = "B"
    elif action == "Move Left":
        location = "A"

    if room_status["A"] == "Clean" and room_status["B"] == "Clean":
        print("\nBoth rooms are clean!")
        break
```
